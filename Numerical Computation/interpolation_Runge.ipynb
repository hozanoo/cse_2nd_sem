# Runge example

The notebook interpolation_simple.ipynb suggests that the interpolation error gets better, if more points are used. 

This is in fact **not always true**!!

In the following, we take the function 
$$
  f(x) = \frac{1}{1+25x^2}
$$
and compute the interpolation polynomials using equidistant and Chebyshev points.

```python
import numpy as np
import matplotlib.pyplot as plt
f = lambda x: 1 / (1 + 25 * x**2)  # function
z = np.linspace(-1, 1, 2001)           # nodes for plotting
```

```python
n = 2  # number nodes for interpolation
x_equi = np.linspace(-1, 1, n + 1)  # equidistant nodes in [-1,1]
p = np.polyfit(x_equi, f(x_equi), n)
p_equi = np.polyval(p, z)
err_equi = np.max(np.abs(f(z) - p_equi))

x_cheb = np.cos((2 * np.arange(n + 1) + 1) / (2 * n + 2) * np.pi)
p = np.polyfit(x_cheb, f(x_cheb), n)
p_cheb = np.polyval(p, z)
err_cheb = np.max(np.abs(f(z) - p_cheb))

plt.figure()
plt.plot(z, p_cheb, '-b', z, p_equi, '-r', z, f(z), '--k', linewidth=2)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.legend(['Cheb.', 'Equi', 'f(x)'])
```

```python
n = 12  # number nodes for interpolation
x_equi = np.linspace(-1, 1, n + 1)  # equidistant nodes in [-1,1]
p = np.polyfit(x_equi, f(x_equi), n)
p_equi = np.polyval(p, z)
err_equi = np.max(np.abs(f(z) - p_equi))

x_cheb = np.cos((2 * np.arange(n + 1) + 1) / (2 * n + 2) * np.pi)
p = np.polyfit(x_cheb, f(x_cheb), n)
p_cheb = np.polyval(p, z)
err_cheb = np.max(np.abs(f(z) - p_cheb))

plt.figure()
plt.plot(z, p_cheb, '-b', z, p_equi, '-r', z, f(z), '--k', linewidth=2)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.legend(['Cheb.', 'Equi', 'f(x)'])
```

```python
n = 22  # number nodes for interpolation
x_equi = np.linspace(-1, 1, n + 1)  # equidistant nodes in [-1,1]
p = np.polyfit(x_equi, f(x_equi), n)
p_equi = np.polyval(p, z)
err_equi = np.max(np.abs(f(z) - p_equi))

x_cheb = np.cos((2 * np.arange(n + 1) + 1) / (2 * n + 2) * np.pi)
p = np.polyfit(x_cheb, f(x_cheb), n)
p_cheb = np.polyval(p, z)
err_cheb = np.max(np.abs(f(z) - p_cheb))

plt.figure()
plt.plot(z, p_cheb, '-b', z, p_equi, '-r', z, f(z), '--k', linewidth=2)
plt.xticks(fontsize=18)
plt.yticks(fontsize=18)
plt.legend(['Cheb.', 'Equi', 'f(x)'])
```

Rather than getting closer to the function f, the interpolating polynomial does approximate the given function worse for growing numbers of uniformly interpolation points.

For Chebyshev points this is not observed, here, increasing the number of points produces better approximations.
